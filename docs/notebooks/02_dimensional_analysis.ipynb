# 2 · Dimensional analysis

A `Dimension` is an **exact** vector of `Fraction` exponents over the seven SI base dimensions. `duq.analysis` names dimensions, decomposes them, and searches for equivalent compositions.

```python
import duq
from fractions import Fraction

energy = duq.dimension("M.L^2.T^-2")
print(energy)
print(duq.dimension("force") == duq.dimension("M.L.T^-2"))
print(duq.dimension("L^3/2").exponents["L"])   # exact, not 1.4999...
```

## Naming and decomposition

```python
from duq.analysis import name_of, decompose

print(name_of(duq.dimension("M.L^2.T^-2")))
print(name_of(duq.dimension("M^2.L")))
decompose(duq.dimension("energy"))
```

## Equivalent compositions

`compositions` rewrites a dimension in terms of named dimensions, shortest first.

```python
from duq.analysis import compositions

comps = compositions(duq.dimension("energy"), max_terms=2)
print({"force": Fraction(1), "length": Fraction(1)} in comps)
comps[0]
```

## Shortest composition — exact or nothing

`shortest_composition` returns the fewest-term exact composition, or **raises** — it never returns a silently-wrong answer.

```python
from duq.analysis import shortest_composition

print(shortest_composition(duq.dimension("energy")))
print(shortest_composition(duq.dimension("force") * duq.dimension("velocity")))
```

```python
shortest_composition(duq.dimension("L^1/2"))
```
